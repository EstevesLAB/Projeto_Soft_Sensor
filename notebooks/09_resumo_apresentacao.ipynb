```python
from pathlib import Path
import sys, warnings
warnings.filterwarnings('ignore')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Localiza a raiz do projeto independentemente de o notebook ser aberto pela pasta raiz ou notebooks.
CANDIDATOS=[Path.cwd(), Path.cwd().parent]
ROOT=next((p.resolve() for p in CANDIDATOS if (p/'dados').exists() and (p/'notebooks').exists()),None)
if ROOT is None: raise FileNotFoundError('Abra o VS Code na pasta projeto_soft_sensor.')
if str(ROOT) not in sys.path: sys.path.insert(0,str(ROOT))
from src.funcoes import *
sns.set_theme(style='whitegrid')
print('Raiz:',ROOT)
```

## Carregamento dos resultados

```python
dados=carregar_csv(ROOT/'dados/03_processados/base_avaliacao_integrada.csv'); ms=pd.read_csv(ROOT/'resultados/metricas/metricas_soft_sensor.csv'); mc=pd.read_csv(ROOT/'resultados/metricas/metricas_classificador.csv',index_col=0); ai=pd.read_csv(ROOT/'resultados/metricas/avaliacao_integrada.csv'); display(ms); display(mc); display(ai)
```

## Painel executivo

```python
fig,axs=plt.subplots(2,2,figsize=(15,9))
axs[0,0].plot(dados.timestamp,dados.temperatura_ambiente_C,label='Externa',lw=.6); axs[0,0].plot(dados.timestamp,dados.temperatura_validada_C,label='Interna validada',lw=.6); axs[0,0].legend(); axs[0,0].set_title('Temperaturas - 30 dias')
dados.classe_sistema.value_counts().plot.bar(ax=axs[0,1]); axs[0,1].set_title('Condições identificadas')
axs[1,0].hist((dados.temperatura_validada_C-dados.temperatura_soft_sensor_C).dropna(),bins=45); axs[1,0].set_title('Diferença: validada - soft sensor')
axs[1,1].axis('off'); texto=f"Soft sensor: {ms.iloc[0]['modelo']}\nMAE: {ms.iloc[0]['MAE']:.3f} °C\nRMSE: {ms.iloc[0]['RMSE']:.3f} °C\nR²: {ms.iloc[0]['R2']:.3f}\nReconstruídos: {int(ai.iloc[0]['reconstruidos'])}"; axs[1,1].text(.05,.85,texto,va='top',fontsize=15,bbox=dict(boxstyle='round',facecolor='#eaf2f8'))
fig.tight_layout(); fig.savefig(ROOT/'resultados/figuras/apresentacao/painel_executivo.png',dpi=180); plt.show()
```

## Tabela final e conclusões

```python
resumo=pd.DataFrame({'indicador':['Registros totais','Percentual reconstruído','MAE soft sensor','RMSE soft sensor','R2 soft sensor'],'valor':[len(dados),ai.iloc[0].pct_reconstruido,ms.iloc[0].MAE,ms.iloc[0].RMSE,ms.iloc[0].R2]}); display(resumo); resumo.to_csv(ROOT/'resultados/tabelas/resumo_apresentacao.csv',index=False,encoding='utf-8-sig')
print('Mensagens para apresentação:')
print('1. Regras físicas tratam impossibilidades e comunicação antes do ML.')
print('2. O classificador é dedicado a travamento e ruído.')
print('3. O soft sensor preserva continuidade e mantém rastreabilidade.')
print('4. Resultados são didáticos e baseados em dados sintéticos de 30 dias.')
```
